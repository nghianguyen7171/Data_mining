## Part B — Foundations: variables, list, dict

### B1–B4 Follow along


```python
# =============================================================================
# B1. Variables & types
# =============================================================================
# A variable is a name bound to a value in memory (think of a labeled box).
# The = operator is assignment (not equality comparison).
#
# Basic types in this example:
# - int   : whole number (25000)
# - float : number with a fractional part (0.1)
# - str   : text in quotes ("Milk tea")
# - bool  : True / False
#
# type(x) returns the class of x — useful when inspecting unfamiliar data.
# =============================================================================

price = 25000          # int: selling price (VND)
tax_rate = 0.1         # float: 10% tax rate
item_name = "Milk tea" # str: product name
in_stock = True        # bool: whether the item is available

print(type(price), type(tax_rate), type(item_name), type(in_stock))

# Price after tax = base price * (1 + tax rate)
# Example: 25000 * 1.1 = 27500.0 (float result because we multiply by 0.1)
price_after_tax = price * (1 + tax_rate)
print(price_after_tax)   # 27500.0
```

```python
# =============================================================================
# B2. List
# =============================================================================
# A list is an ORDERED, MUTABLE collection.
# Written with [], elements separated by commas.
#
# len(list)      : number of elements
# list.append(x) : add x at the END (modifies the list in place)
# sum(list)      : sum of numeric elements
# max(list)      : largest element
# =============================================================================

revenue = [120, 150, 90, 200, 175]   # daily revenue for 5 days (million VND)
print(len(revenue))    # 5 — five days so far
revenue.append(210)    # add day 6; list length becomes 6
print(revenue)
print(sum(revenue))    # total revenue over the (now) 6 days
print(max(revenue))    # best day
```

```python
# =============================================================================
# B3. Dict — key → value mapping
# =============================================================================
# A dict is {key: value}. Access by KEY NAME, not by numeric position.
# product['price'] reads the value for key 'price'.
# Assigning product['discount'] = 0.15 ADDS a new key if missing, or overwrites it.
# =============================================================================

product = {'name': 'Latte', 'price': 45000, 'stock': 30}
print(product['price'])           # 45000
product['discount'] = 0.15        # add a new key–value pair
print(product)
```

```python
# =============================================================================
# B4. None — the "empty / not yet set" value
# =============================================================================
# None means "no value".
# Preferred check: use "is None" (Python style; avoid "== None").
# In real data, blanks often become None (Python objects) or np.nan (NumPy/pandas).
# =============================================================================

rating = None
print(rating is None)   # True — no rating yet
```

### ✅ Try yourself B — Core (1–6) + Challenge (7–9)


```python
# =============================================================================
# EXERCISES B — CORE (1–6): café scenario
# =============================================================================

# ----- Q1 -----
# Build a 4-month cost list, then use sum/max — same idea as revenue in B2.
cost = [30, 45, 25, 60]   # million VND
print("Total cost:", sum(cost))   # 30+45+25+60 = 160
print("Highest month:", max(cost)) # 60

# ----- Q2 -----
# Customer dict: add a new key by assignment customer['city'] = ...
customer = {'name': 'An', 'age': 28}
customer['city'] = 'Hanoi'
print(customer)  # {'name': 'An', 'age': 28, 'city': 'Hanoi'}

# ----- Q3 -----
# Promo code not entered yet → None; check with "is None".
promo_code = None
print("No code yet?", promo_code is None)  # True

# ----- Q4 -----
# Menu is dict name → price. Read Mocha, then add Tea.
menu = {'Latte': 45000, 'Espresso': 35000, 'Mocha': 50000}
print("Mocha price:", menu['Mocha'])  # 50000
menu['Tea'] = 30000
print(menu)

# ----- Q5 -----
# orders is a list of items sold today.
# len(orders) = total cups; orders.count('Latte') = how many times 'Latte' appears.
orders = ['Latte', 'Mocha', 'Latte', 'Tea', 'Latte']
print("Total cups:", len(orders))            # 5
print("Latte cups:", orders.count('Latte'))  # 3

# ----- Q6 -----
# Loop over each item in orders; add menu[item] into a running total.
# Pattern: "look up price in a dict by key" — common in order analytics.
total = 0
for item in orders:
    # item is successively 'Latte', 'Mocha', 'Latte', 'Tea', 'Latte'
    total += menu[item]
print("Revenue:", total)
# 45000*3 + 50000 + 30000 = 135000 + 50000 + 30000 = 215000
```

```python
# =============================================================================
# EXERCISES B — CHALLENGE (7–9): list of dicts (table-like records)
# =============================================================================
# Each element of inventory is ONE record (one row), itself a dict.
# This is the mental bridge to pandas tables later (each row = one observation).
# =============================================================================

inventory = [
    {'sku': 'L01', 'name': 'Latte',    'price': 45000, 'stock': 12},
    {'sku': 'E02', 'name': 'Espresso', 'price': 35000, 'stock': 3},
    {'sku': 'M03', 'name': 'Mocha',    'price': 50000, 'stock': 0},
    {'sku': 'T04', 'name': 'Tea',      'price': 30000, 'stock': 8},
]

# ----- Q7: filter low stock (stock < 5) -----
# Idea: loop each item; if the condition holds, append item['name'] to a new list.
low_stock = []
for item in inventory:
    # item['stock'] is the integer inventory count for the current record
    if item['stock'] < 5:
        low_stock.append(item['name'])
print("Low stock:", low_stock)  # ['Espresso', 'Mocha']  (3 and 0 are both < 5)

# ----- Q8: inventory value = sum(price * stock) -----
# For each item: capital tied in stock = unit price * quantity on hand.
inventory_value = 0
for item in inventory:
    inventory_value += item['price'] * item['stock']
print("Inventory value (before):", inventory_value)
# Latte 45000*12=540000; Espresso 35000*3=105000; Mocha 0; Tea 30000*8=240000
# Total = 540000+105000+240000 = 885000

# ----- Q9: receive 10 more Espresso units, then recompute -----
# Find the record whose name == 'Espresso', then increase stock.
# break: stop the loop once found (no need to scan further).
for item in inventory:
    if item['name'] == 'Espresso':
        item['stock'] += 10   # 3 + 10 = 13
        break

# Recompute inventory value after the delivery
inventory_value = 0
for item in inventory:
    inventory_value += item['price'] * item['stock']
print("Inventory value (after Espresso delivery):", inventory_value)
# Increases by 35000*10 = 350000 → 885000 + 350000 = 1235000
```

## Part C — Indexing & slicing (focus)

### C1–C3 Follow along


```python
# =============================================================================
# C1. Indexing — get EXACTLY ONE element
# =============================================================================
# Python indices START AT 0:
#   position: 0          1       2            3       4
#   value:    Espresso   Latte   Cappuccino   Mocha   Americano
#
# NEGATIVE indices count from the end: [-1] = last, [-2] = second-to-last, ...
# Strings (str) are indexed character-by-character with the same rules.
# =============================================================================

drinks = ['Espresso', 'Latte', 'Cappuccino', 'Mocha', 'Americano']
print(drinks[0])    # 'Espresso' — first
print(drinks[2])    # 'Cappuccino'
print(drinks[-1])   # 'Americano' — last
print(drinks[-2])   # 'Mocha'

code = "SP2025"
print(code[0])      # 'S'
print(code[-1])     # '5'
```

```python
# =============================================================================
# C2. Slicing — get a SEGMENT: x[start:stop:step]
# =============================================================================
# GOLDEN RULE: take from start up to, but NOT including, stop.
# This is the #1 student mistake.
#
# Omit start → default from the beginning
# Omit stop  → default through the end
# step = 2   → take every 2nd element
# step = -1  → walk backwards (reverse a list/string)
# =============================================================================

prices = [20, 35, 45, 60, 80, 95]
# index:   0   1   2   3   4   5

print(prices[1:4])   # [35, 45, 60] — indices 1,2,3; NOT 4
print(prices[:3])    # [20, 35, 45] — from start up to before 3
print(prices[3:])    # [60, 80, 95] — from 3 to the end
print(prices[::2])   # [20, 45, 80] — step 2
print(prices[::-1])  # full reverse
```

```python
# =============================================================================
# C3. Small application: split weekdays / weekend with slicing
# =============================================================================
# week_rev[0] = Mon ... week_rev[6] = Sun
# [:5]  = Mon–Fri (first 5 days)
# [-2:] = last 2 elements = Sat, Sun
# =============================================================================

week_rev = [120, 150, 90, 200, 175, 210, 160]
weekdays = week_rev[:5]
weekend = week_rev[-2:]
print("Weekdays:", weekdays)
print("Weekend:", weekend)
```

## Part D — Minimal NumPy

### D1–D3 Follow along


```python
# =============================================================================
# D1. NumPy arrays & indexing/slicing
# =============================================================================
# np.array([...]) builds a numeric array. 1D indexing/slicing works LIKE lists.
# 2D arrays: think of a table — [row, column], still 0-based.
# A lone : means "take all" along that axis.
# =============================================================================

import numpy as np

a = np.array([10, 20, 30, 40, 50])
print(a[0])      # 10
print(a[1:3])    # array([20, 30]) — stop is exclusive
print(a[-1])     # 50

m = np.array([[1, 2, 3],
              [4, 5, 6]])
#       col0 col1 col2
# row0   1    2    3
# row1   4    5    6
print(m[0, 1])   # 2 — row 0, column 1
print(m[:, 0])   # [1, 4] — all rows, column 0
print(m[1, :])   # [4, 5, 6] — row 1, all columns
```

```python
# =============================================================================
# D2. Vectorization — compute on the WHOLE array at once (no for-loop)
# =============================================================================
# revenue * 1.1 multiplies EACH element by 1.1.
# This is why NumPy/pandas is fast and more concise than pure Python loops.
# .mean() / .sum() are statistical methods on the array.
# =============================================================================

revenue = np.array([120, 150, 90, 200])
print(revenue * 1.1)
print(revenue - 50)
print(revenue.mean())
print(revenue.sum())
```

```python
# =============================================================================
# D3. np.nan — representing MISSING values
# =============================================================================
# nan = "Not a Number", but in analytics it marks an empty / missing cell.
# np.isnan(array) → boolean array True where values are missing.
# np.isnan(...).sum() counts missing cells (True counts as 1).
# np.nanmean(...) averages while IGNORING nan (plain .mean() becomes nan if any nan).
# =============================================================================

b = np.array([100, 120, np.nan, 95])
print(b)
print(np.isnan(b))
print("Missing count:", np.isnan(b).sum())
print("Mean ignoring nan:", np.nanmean(b))
```

```python
# =============================================================================
# EXERCISES D — CHALLENGE (6–9): clean, then analyse
# =============================================================================
# Typical workflow: detect missing → estimate (nanmean) → impute → analyse.
# =============================================================================

import numpy as np

sales = np.array([200., 240., np.nan, 300., 180., 260.])
tbl = np.array([
    [10., 20., 15.],
    [30., 40., 35.],
    [50., 60., 55.],
    [25., np.nan, 40.],
])  # 4 stores × 3 quarters

# ----- Q6 -----
print("nan count in sales:", np.isnan(sales).sum())  # 1
mu = np.nanmean(sales)
print("nanmean sales:", mu)  # 236.0
# (200+240+300+180+260)/5 = 1180/5 = 236 — the missing month is excluded

# ----- Q7: impute — replace nan with the mean -----
# .copy() so we do NOT overwrite the original array (keep sales for comparison).
# Boolean mask: sales_filled[np.isnan(sales_filled)] selects only the nan cells.
sales_filled = sales.copy()
sales_filled[np.isnan(sales_filled)] = mu
print("After fill:", sales_filled)
print("Mean after fill:", sales_filled.mean())  # also 236.0

# ----- Q8: slice the 2D table -----
print("Row 0 (store 1):", tbl[0, :])
print("Col 1 (Q2):", tbl[:, 1])      # includes nan for store 4
print("Store3-Q3:", tbl[2, 2])       # 55.0

# ----- Q9 -----
print("nan count in tbl:", np.isnan(tbl).sum())  # 1
tbl_q1 = tbl[:, 0]
print("Q1 sum:", tbl_q1.sum())  # 10+30+50+25 = 115
# Missing cell: row 3 (store 4), column 1 (quarter 2) — see np.isnan(tbl)
print("Missing mask:")
print(np.isnan(tbl))
```

## Part E — Integrated mini-case (required)

Wednesday revenue is missing → impute before computing profit / margin.


```python
# =============================================================================
# MINI-CASE — Input data
# =============================================================================
# days / revenue / cost are THREE PARALLEL arrays aligned by day order.
# store is a nested dict: open_hours is itself a dict.
# =============================================================================

import numpy as np

days = np.array(['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])
revenue = np.array([120., 150., np.nan, 200., 175., 210., 160.])  # million VND
cost = np.array([80., 95., 70., 110., 100., 120., 90.])
store = {
    'name': 'Store A',
    'region': 'Hanoi',
    'staff': 8,
    'open_hours': {'weekday': 10, 'weekend': 12},
}
```

```python
# =============================================================================
# Mini-case — Q1 to Q4
# =============================================================================

# Q1: nested dict access
# store['open_hours'] returns the inner dict; ['weekday'] then reads opening hours.
print(
    store['name'],
    store['region'],
    store['open_hours']['weekday'],
)

# Q2: which day name(s) have missing revenue?
# np.isnan(revenue) → bool mask; days[mask] keeps only matching day names.
print("Days with missing revenue:", days[np.isnan(revenue)])  # ['Wed']

# Q3: impute with nanmean
# Copy, then set nan cells = mean of the observed days.
revenue_filled = revenue.copy()
revenue_filled[np.isnan(revenue_filled)] = np.nanmean(revenue)
print("Revenue after fill:", revenue_filled)
# nanmean ≈ (120+150+200+175+210+160)/6 = 1015/6 ≈ 169.1667

# Q4: weekday / weekend totals via slicing on the filled array
weekday_rev = revenue_filled[:5].sum()
weekend_rev = revenue_filled[-2:].sum()
print("Weekday revenue total:", weekday_rev)
print("Weekend revenue total:", weekend_rev)
```

```python
# =============================================================================
# Mini-case — Q5 to Q9
# =============================================================================

# Q5: daily profit = revenue - cost (element-wise vectorization)
profit = revenue_filled - cost
print("Profit:", profit)

# Q6
print("Average profit:", profit.mean())

# Q7: day NAMES whose profit is strictly above the mean
# Comparison builds a mask; use it to select from days.
print("Days with profit > mean:", days[profit > profit.mean()])

# Q8: day with the highest profit
# np.argmax returns the INDEX (int) of the max — then look up days[index].
best_idx = np.argmax(profit)
print("Best profit day:", days[best_idx], "with", profit[best_idx])
# Note: Wednesday was filled with a fairly high mean revenue while Wed cost is low (70),
# so Wed profit can become the highest — that is the effect of imputation.

# Q9: profit margin = profit / revenue
# Element-wise division; then compare mean weekday vs weekend margin.
margin = profit / revenue_filled
weekday_m = margin[:5].mean()
weekend_m = margin[-2:].mean()
print("Mean weekday margin:", weekday_m)
print("Mean weekend margin:", weekend_m)
if weekday_m > weekend_m:
    print("→ Weekday margin is higher")
elif weekend_m > weekday_m:
    print("→ Weekend margin is higher")
else:
    print("→ Approximately equal")
```
